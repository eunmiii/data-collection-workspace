***깔끔한 데이터(tidy data)***
1. R커뮤니티의 해드리 위컴이  한 논문에서 처음 소개한 개념으로,데이터셋을 구조화하는 프레임워크이다.    

2. 해드리 위컴이 논문에서 제시한 깔끔한 데이터의 조건

   (1) 행은 관측값으로 나타내야 한다.   
   (2) 열은 변수를 나타내야 한다.   
   (3) 관측 단위별로 데이터 표를 구성해야 한다.  
3. 해드리 위컴이 저서 <R을 활용한 데이터 과학>에서 하나의 데이터셋, 즉 표에 초점을 두고 깔끔한 데이터 정의를 했다.  
   - 변수는 열로 나타내야 한다.  
   - 관측값은 행으로 나타내야 한다.
   - 값은 셀로 나타내냐 한다.

4. pandas에서 데이터프레임을 깔끔한 데이터로 정리하는데 유용한 melt메서드를 제공한다.


   |메서드 인자|설명|
   |---|---|
   |id_vars| 위치를 그대로 유지할 열의 이름 지정한다.|
   |value_vars|행으로 위치를 변경할 열의 이름을 지정한다. 기본적으로 id_vars로 지정하지 않은 모든 열을 대상으로 설정된다.|
   |var_name|value_vars로 위치를 변경한 열의 이름을 지정한다. 기본값은 'variable' 이다.|
   |value_name|var_name으로 위치를 변경한 열의 데이터를 지정한 열의 이름을 지정한다. 기본값은 'value'이다.|

```python
# -----------------------------
# 1) 최신 2개 연도 추출
# -----------------------------
recent_years = sorted(df_local["연도"].unique())[-2:]
print("최근 2년:", recent_years)

df_recent = df[df["연도"].isin(recent_years)]

# -----------------------------
# 2) 연도별 전체 출생아수 합계 계산
# -----------------------------
total_birth_by_year = df_recent.groupby("연도")["출생아수"].sum()

# -----------------------------
# 3) 시군구별 출생아 비율(%) 계산
# -----------------------------
df_recent["출생아_비율(%)"] = df_recent.apply(
    lambda x: (x["출생아수"] / total_birth_by_year[x["연도"]]) * 100,
    axis=1
)

# -----------------------------
# 4) 결과 출력
# -----------------------------
result = df_recent[["연도", "시군구별", "출생아수", "출생아_비율(%)"]].sort_values(["연도", "출생아_비율(%)"], ascending=[False, False])

display(result)
```

```python


# -----------------------------
# 1) 최신 2개 연도 추출
# -----------------------------
year_range = df_local_total["연도"].sort_values().unique()[-2:]
df_recent = df_local_total[df_local_total["연도"].isin(year_range)].copy()

print("분석 대상 연도:", year_range)


# -----------------------------
# 2) 지역별(시군구별) 연도별 출생아수 합계
# -----------------------------
group_sum = df_recent.groupby(["시군구별", "연도"])["출생아수"].sum().reset_index()
print(group_sum)

# -----------------------------
# 3) Pivot 변환 (연도를 컬럼으로)
# -----------------------------
pivot_df = group_sum.pivot(index="시군구별", columns="연도", values="출생아수")


# -----------------------------
# 4) 2023 대비 2024 증감률(%) 계산
# -----------------------------

y0, y1 = year_range[0], year_range[1]   # '2023', '2024'

pivot_df["증감률(%)"] = np.where(
    pivot_df[y0] == 0,
    np.nan,
    ((pivot_df[y1] - pivot_df[y0]) / pivot_df[y0]) * 100
)



# -----------------------------
# 5) 증감률 기준 내림차순 정렬
# -----------------------------
pivot_df = pivot_df.sort_values("증감률(%)", ascending=False)

print(pivot_df)
```

```python


# -----------------------------
# 1) pointplot용 데이터프레임 변환
# -----------------------------
plot_df = pivot_df.reset_index()[["시군구별", "증감률(%)"]].dropna()



# 최대/최소 값 찾기
max_idx = plot_df["증감률(%)"].idxmax()
min_idx = plot_df["증감률(%)"].idxmin()

max_value = plot_df.loc[max_idx, "증감률(%)"]
min_value = plot_df.loc[min_idx, "증감률(%)"]

# -----------------------------
# 2) 시군구별 증감률(%) pointplot
# -----------------------------
ax = plt.figure(figsize=(12, 6))



ax = sns.pointplot(
    data=plot_df,
    x="시군구별",
    y="증감률(%)"
)

plt.title("전녀대비 증가율(%)")
plt.xlabel("시군구별")
plt.ylabel("증감률(%)")
plt.xticks(rotation=45)
plt.grid(False)

# -----------------------------
# 3) 라벨 표시 + 최대/최소 강조
# -----------------------------
for i, row in enumerate(plot_df.itertuples()):
    value = row._2   # 증감률(%)
    region = row.시군구별

    color = "black"
    if value == max_value:
        color = "red"
    elif value == min_value:
        color = "green"

    ax.text(i, value+0.2, f"{value:.2f}%", ha="center", va="bottom", fontsize=10, color=color)

# -----------------------------
# 4) 최대/최소 점을 scatter로 다시 찍어서 강조
# -----------------------------
max_pos = plot_df[plot_df["증감률(%)"] == max_value].index[0]
min_pos = plot_df[plot_df["증감률(%)"] == min_value].index[0]

ax.scatter(plot_df.index.get_loc(max_pos), max_value, color="red", s=120, zorder=5)
ax.scatter(plot_df.index.get_loc(min_pos), min_value, color="green", s=120, zorder=5)


plt.show()
```
