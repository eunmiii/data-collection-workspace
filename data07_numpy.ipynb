## np.expand_dims(a, axis) = 명시적 차원 추가

```python
import numpy as np

arr = np.array([1, 2, 3]) # shape: (3,)

# axis=0 위치에 차원 추가 -> shape: (1, 3)
e0 = np.expand_dims(arr, axis=0)

# axis=1 위치에 차원 추가 -> shape: (3, 1)
e1 = np.expand_dims(arr, axis=1)

print("e0.shape:", e0.shape) # (1, 3)
print("e1.shape:", e1.shape) # (3, 1)
```

## 백터화 연산과 브로드캐스팅  
- 작은걸 큰거에 맞춤

 **4) 암기 꿀팁 (헷갈릴 때 기억하는 방법)"지우개 법칙 (축을 없앤다!)"**

- `shape (3, 3)` 형태의 2차원 배열이 있을 때:
  - `axis=0`으로 연산하면 (3, 3)중  첫 번째 3(행 축)이 지워져 결과는 **`shape (3,)`** (세로 연산 결과)가 된다.
  - `axis=1`로 연산하면 (3, 3) 중 두 번째 3(열 축)이 지워져 결과는 **`shape (3,)`** (가로 연산 결과)가 된다.

즉, **`axis=n`은 "N번째 차원 방향으로 데이터들을 뭉개서 하나로 합친다!"** 라고 이해하시면 가장 쉽다.

```python
import numpy as np

# 1. 백터 및 행렬 생성
v1 = np.array([1,2,3])
v2 = np.array([4,5,6])

A = np.array([
    [1,2],
    [3,4]
])


B = np.array([
    [5,6],
    [7,8]
])


# 2.백터 내적 (Dot Product)
dot_product = np.dot(v1,v2)
print(f'Vector Dot Product : {dot_product}')
```

```python
# 3.행렬 곱셈 (Matrix Multipliecation)
matrix_mul = np.matmul(A,B)
print('Matrix Multiplication (A @ B) :')

print(matrix_mul)
```

```python
# 4.전치 행렬 (Transpose Matrix)
A_transpose = A.T
print('transpose of Matrix A : ')
print(A_transpose)
```

배열은 @가능하지만 *은 사용 할 수 없다.

matrix는 * 와 @ 모두가 가능하다.

```python
c = np.matrix([[1,2], [3,4]]) # 행렬 생성
print(c)

print(c)
print(c.shape)

print(c * c.T)
print(c * np.dot(c.T))  # 함수방법으로 np.dot( )함수를 사용할 수 있다.
print( c.dot(c.T) ) # 메소드(method) 방식
print(c @ c.T)

'''
matrix([[ 5, 11],
        [11, 25]])
'''
```

```python
import numpy as np

# 1. 시드 고정 및 가상의 학생 성적 데이터셋 생성 (학생 5명, 과목 3개)
np.random.seed(42)
scores = np.random.randint(40, 101, size=(5, 3))

print("1. 원본 성적 데이터셋 (5명 x 3과목):")
print(scores)

# 2. 각 과목별 평균 및 표준편차 구하기 (axis=0: 세로/열 방향 연산)
subject_mean = np.mean(scores, axis=0)
subject_std = np.std(scores, axis=0)

print("\n2. 과목별 평균:")
print(subject_mean)

print("\n3. 과목별 표준편차:")
print(np.round(subject_std, 2))

# 3. 각 학생별 평균 구하기 (axis=1: 가로/행 방향 연산)
student_mean = np.mean(scores, axis=1)

print("\n4. 학생별 평균 점수:")
print(np.round(student_mean, 2))

# 4. 낙제점 수정하기 (60점 미만인 점수를 60점으로 일괄 수정)
# np.where(조건, 참일 때 값, 거짓일 때 값) 활용
fixed_scores = np.where(scores < 60, 60, scores)

print("\n5. 낙제점(60점 미만) 수정 후 데이터셋:")
print(fixed_scores)
```

```python
import numpy as np

# 1. 가상의 RGB 이미지 픽셀 데이터 생성 (높이 4 x 너비 4 x 채널 3)
np.random.seed(100)
image_rgb = np.random.randint(0, 256, size=(4, 4, 3), dtype=np.uint8)

print("1. 원본 RGB 이미지 Shape:")
print(image_rgb.shape)

# 2. RGB 흑백 전환 (Grayscale Conversion)
# ITU-R BT.601 표준 공식: Y = 0.299*R + 0.587*G + 0.114*B
# 채널 축(axis 2)에 가중치를 내적(Dot Product)하거나 브로드캐스팅하여 계산함
weights = np.array([0.299, 0.587, 0.114])
image_gray = np.dot(image_rgb, weights).astype(np.uint8)

print("\n2. 흑백 전환 후 이미지 Shape:")
print(image_gray.shape)  # (4, 4)로 차원 축소됨

print("\n3. 흑백 이미지 픽셀 값:")
print(image_gray)

# 3. 이미지 크기 조절 (Downsampling / Resizing)
# 슬라이싱([::2, ::2])을 이용하여 가로/세로 픽셀을 2칸씩 스킵함으로써 크기를 1/2로 줄임
image_resized = image_gray[::2, ::2]

print("\n4. 크기 조절(2x2로 축소) 후 이미지 Shape:")
print(image_resized.shape)

print("\n5. 최종 축소된 이미지 픽셀 값:")
print(image_resized)
```
