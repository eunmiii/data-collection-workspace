```python
import numpy as np
```

```python
data = np.arange(10)
print(data)
```

```python
item = np.random.shuffle(data)
print('item:', item)
print('data:', data)
```

```python
data = np.arange(10)
print(data)
```

```python
item = np.random.permutation(data)
print('data', data)
print('item', item)
```
