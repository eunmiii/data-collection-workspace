# Seaborn

### 1. Seaborn이란? (정의)
- Seaborn은 Matplotlib을 기반으로 구축된 파이썬 고수준 통계 데이터 시각화 라이브러리이다.

- Pandas의 데이터프레임(DataFrame) 구조와 완벽하게 통합되어 있어, 복잡한 통계 수치나 데이터 분포를 쉽고 아름답게 시각화할 수 있도록 도와준다.

### 2. 주요 특징 (Features)
- 아름다운 기본 디자인: 별도의 세밀한 설정 없이도 세련되고 정돈된 팔레트와 스타일을 기본으로 제공한다.

- Pandas 데이터프레임 최적화: data=df, x='col1', y='col2', hue='category' 형태의 컬럼명 매핑 방식으로 코드가 직관적이다.

- 통계 연산 기능 내장: 단순 데이터 출력뿐만 아니라 평균, 중앙값, 신뢰구간(CI), 회귀 추세선, 커널 밀도 추정(KDE) 등을 자동으로 계산하여 그려준다.

- 다차원(카테고리별) 분석 용이 (hue 파라미터): 단 한 줄의 hue 옵션 추가만으로 범주별 색상 구분이 가능하다.

- 서브플롯 격자 자동화 (FacetGrid): 조건(범주)에 따라 여러 개의 그래프를 격자(Grid) 형태로 나누어 그리는 기능이 강력하다.

### 3. 핵심 기능 및 차트 종류

(1) Seaborn 시각화 목적
```
 ├── 1. 시간 흐름/추세 파악 ───► 선그래프 (lineplot), 회귀선 추세 (regplot, lmplot)
 ├── 2. 항목 간 크기 비교 ────► 막대그래프 (barplot), 카운트플롯 (countplot), 점 그래프 (pointplot)
 ├── 3. 변수 간 상관관계 ────► 산점도 (scatterplot), 히트맵 (heatmap), 변수쌍 차트 (pairplot)
 ├── 4. 분포 및 이상치 확인 ──► 히스토그램 (histplot), 밀도 그래프 (kdeplot), 바이올린플롯 (violinplot), 스트립/스웜플롯 (stripplot, swarmplot)
 ├── 5. 범주/다차원 분석 ────► 범주형 타일링 (catplot), 격자형 조합 (FacetGrid, JointGrid)
 └── 6. 전체 대비 비율 확인 ──► 누적 막대그래프 (barplot with native scaling)
```
(2) 주요 메서드

| **목적** | **주요 함수 / 그래프 종류** | **설명** |
| --- | --- | --- |
| **추세 파악** | `sns.lineplot()`, `sns.regplot()`, `sns.lmplot()` | 오차범위(신뢰구간)가 자동 표시되는 선 그래프 및 선형회귀 추세선 |
| **크기 비교** | `sns.barplot()`, `sns.countplot()`, `sns.pointplot()` | 평균/신뢰구간이 자동 계산되는 막대, 범주별 데이터 개수(Count) 전용 그래프 |
| **상관 관계** | `sns.scatterplot()`, `sns.heatmap()`, `sns.pairplot()` | 색상/크기 매핑 산점도, 상관계수 행렬 히트맵, 모든 변수 간 1:1 관계 일괄 출력 |
| **분포/이상치** | `sns.histplot()`, `sns.kdeplot()`, `sns.stripplot()`, `sns.swarmplot()` | 커널밀도추정(KDE) 곡선, 점의 겹침을 방지하여 데이터 분포를 그대로 보여주는 Swarm Plot |
| **다차원 분석** | `sns.catplot()`, `sns.displot()`, `sns.FacetGrid()` | 조건에 따라 여러 개의 서브플롯을 격자형(Grid)으로 자동 나열 |

### 4. 장점 및 단점
⭕ 장점
- 적은 코드량: Matplotlib으로 수십 줄 써야 하는 통계 그래프를 1~2줄로 완성할 수 있다.  
- EDA(탐색적 데이터 분석) 속도 향상: 데이터 탐색 단계에서 그룹별 비교가 매우 빨라진다.  
- 통계적 신뢰성 제공: 신뢰구간이나 밀도 추정이 기본 포함되어 통계적 통찰을 빠르게 얻을 수 있다.  

❌ 단점
- 커스텀의 한계: 복잡하고 독창적인 형태의 커스텀 차트를 만들 때는 제약이 있다.  

- Matplotlib 의존성: 세부적인 폰트, 축 위치, 타이틀 세팅 등을 완벽하게 조절하려면 결국 Matplotlib 명령어를 사용해야 한다.

> "Matplotlib의 세밀함에 Pandas의 편리함과 통계적 아름다움을 더한 라이브러리"

> 단순 시각화를 넘어 데이터 간의 통계적 관계나 패턴을 빠르게 파악하고 보고서용 차트를 만들 때 가장 추천하는 라이브러리이다.

### 5. Seaborn v0.14 이후 주요 변경사항 & 대응 가이드

(1) palette 사용 시 hue 명시 필수

```
# 이전 방식 (v0.13 이하까지 허용됨)
sns.barplot(data=df, x='bmi_group', y='target', palette='Blues')

# 앞으로 권장 방식 (v0.14 이상)
sns.barplot(data=df, x='bmi_group', y='target', hue='bmi_group',
            palette='Blues', legend=False)

# 이유: palette는 색상 맵핑을 위해 hue가 필요합니다. hue 없이 쓰면 경고 또는 오류가 발생할 수 있습니다.
```

(2) ci → errorbar로 변경

```
# 이전
sns.barplot(data=df, x='bmi_group', y='target', ci='sd')

# 현재 (v0.13~)
sns.barplot(data=df, x='bmi_group', y='target', errorbar='sd')

이유: ci(confidence interval)는 더 이상 사용되지 않으며, errorbar='sd', errorbar='se' 등으로 대체되었습니다.
sd은 표준편차(standard deviation), se은 표준오차(standard error)
```

## 1.1 막대 그래프

errorbar 옵션

| 목적                       | 선택할 errorbar |
| ------------------------ | ------------ |

| 평균값의 **신뢰도/정확도**를 보고 싶다면 | `'se'`       |
| 오차막대가 필요 없다면             | `None`       |



palette 옵션

| 팔레트 이름                 | 설명              |

| `Reds`                 | 빨간색 계열 밝기 변화    |
| `Greens`               | 초록색 계열 밝기 변화    |
| `Purples`              | 보라색 계열 밝기 변화    |
| `Set1`, `Set2`, `Set3` | 대비가 강한 여러 색상 세트 |
| `coolwarm`             | 차가운 색과 따뜻한 색 조합 |
| `viridis`              | 밝기 순서가 명확한 컬러맵  |

````python

plt.figure(figsize=(8, 5))

# sns.barplot(data=df, x='bmi_group', y='target', hue='bmi_group', palette='Blues', errorbar=None)
# - 각 bmi_group별 target 평균을 막대그래프로 표현
# - hue='bmi_group'로 그룹별 색상을 다르게 지정
# - palette='Blues'를 사용해 파란색 계열 색상 그라데이션 적용
# - errorbar=None 으로 오차막대는 표시하지 않음
# (현재 주석 처리된 상태)

sns.barplot(
    data=df,
    x='bmi_group',
    y='target',
    color='steelblue',  # 모든 막대를 동일한 'steelblue' 색상으로 표시
    errorbar='sd'       # 오차막대 표시 안함.       # sd : 표준편차
)

plt.title("[시본] BMI 그룹별 평균 Target")  # 그래프 제목 설정
plt.show()

# 실행 결과 설명
# 사용 데이터: BMI 그룹(bmi_group)에 따른 당뇨 진행도(target)의 평균값
# 용도: BMI 구간별 당뇨 진행도 평균 비교 및 시각적 표현
# 해석:
# ① BMI가 높을수록 평균 target 값이 증가하는 경향이 있음 → BMI가 높을수록 당뇨 진행 위험도가 커질 가능성 시사
# ② 각 그룹별 막대가 비교적 뚜렷하게 구분되어 BMI 그룹별 차이가 명확함
# ③ errorbar가 없으므로 개별 그룹 내 데이터 분포의 변동성은 표시하지 않음
# ④ 단일 색상 사용 시 시각적 단순화 효과가 있으나, 그룹별 차이를 색상으로 표현하는 것보다는 덜 직관적임
````

## 1.2 산점도 그래프

````python

plt.figure(figsize=(6, 5))

# BMI와 당뇨 진행도(target) 간의 관계를 산점도로 시각화
# hue='bmi_group'으로 BMI 그룹별 색상을 달리하여 시각적 구분 강조
# palette='Set2'는 대비가 좋은 여러 색상 세트를 사용
sns.scatterplot(data=df, x='bmi', y='target', hue='bmi_group', palette='Set2')

plt.title("[시본] BMI와 Target 산점도")  # 그래프 제목 설정
plt.show()


# 실행 결과 설명
# 사용 데이터: 개별 환자의 BMI와 당뇨 진행도(target) 값
# 용도: BMI와 당뇨 진행도 간 상관 관계 탐색 및 그룹별 패턴 비교
# 해석:
# ① BMI가 증가할수록 target 값도 전반적으로 증가하는 경향이 보임 (양의 상관관계)
# ② 색상으로 BMI 그룹(Low, Medium, High)을 구분해 각 그룹 내 분포 차이를 쉽게 파악 가능
# ③ High 그룹은 상대적으로 target 값이 높고 밀집된 영역이 확인되어 위험군 분포 시사
# ④ 산점도 점들의 투명도(기본 alpha값)로 일부 중첩도 표현되어 데이터 밀집도 파악 가능
````

## 1.3 히스토그램 & 커널 밀도 (distplot → displot 권장)

````python
# kde=False 이면, 커널 밀도가 안 보인다. (곡선)
sns.displot(df['target'], kde=True, bins=20, color='tomato')       # Kernel Density Estimation

# displot은 Figure 객체 단위로 플롯을 생성하므로 plt.title() 대신
# FacetGrid 객체의 set_titles()나 set() 메서드를 사용해야 함.
# 하지만 단순히 제목만 추가하려면 다음과 같이 작성 가능:
plt.subplots_adjust(top=0.9)  # 제목 공간 확보
plt.suptitle("[시본] Target 히스토그램 + 밀도곡선")

plt.show()


# 실행 결과 설명
# 사용 데이터: 'target' (당뇨 진행도) 변수 단일 연속형 데이터
# 용도: 데이터 분포의 형태 파악 (빈도, 분포 밀도 포함)
# 해석:
# ① target 값은 대체로 100350 구간에 분포하며, 100200 사이에 관측치가 가장 많음
# ② 히스토그램과 함께 커널 밀도 추정(kde) 곡선을 통해 부드러운 분포 형태를 시각화
# ③ 정규분포 형태를 일부 보이나 약간 오른쪽으로 치우침(우측 꼬리 현상) 가능성
# ④ 데이터 분포에 대한 이해를 바탕으로 통계 분석 및 모델링 전처리 전략 수립 가능
````

## 1.4 박스플롯

````python
plt.figure(figsize=(6, 4))

# BMI 그룹별 당뇨 진행도(target)의 분포를 박스플롯으로 시각화
# hue='bmi_group'과 x='bmi_group'이 동일해 중복된 그룹 표시가 되어 다소 복잡할 수 있음
# palette='pastel'은 부드러운 파스텔톤 색상을 사용
# orientation='vertical'은 기본값으로 세로 방향 박스플롯 지정
sns.boxplot(data=df, x='bmi_group', y='target', hue='bmi_group', palette='pastel', orientation='vertical')

plt.title("[시본] BMI 그룹별 박스플롯")
plt.show()


# 실행 결과 설명
# 사용 데이터: BMI 그룹에 따른 당뇨 진행도 분포
# 용도: 중앙값, 사분위 범위, 이상치 확인으로 그룹별 데이터 분포 특성 분석
# 해석:
# ① 각 BMI 그룹별 중앙값과 사분위 범위가 차이를 보이며, BMI가 높을수록 target 값이 높은 경향
# ② 동일한 그룹으로 hue를 지정해 색상이 중복되어 시각적으로 다소 복잡해질 수 있음
# ③ 이상치(아웃라이어)가 각 그룹에 분포하는지 여부 확인 가능
# ④ 그룹 간 분포 차이 및 변동성 파악에 용이하며, BMI 그룹에 따른 당뇨 위험도 비교에 활용 가능
````

````python
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, y='bmi_group', x='target', hue='bmi_group', palette='pastel', orientation='horizontal')

plt.title("[시본] BMI 그룹별 박스플롯")
plt.show()
````

## 1.5 바이올린 플롯

## 1.7 히트맵 (상관계수 시각화)

````python
import numpy as np

# 숫자 타입 컬럼만 선택해 상관계수 계산
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()  # 수치형 컬럼 간 피어슨 상관계수 계산

plt.figure(figsize=(10, 6))

# 상관계수 행렬을 heatmap으로 시각화
# annot=True: 각 셀에 상관계수 숫자 표시
# cmap='coolwarm': 양수는 붉은색 계열, 음수는 파란색 계열로 색 구분
# fmt=".2f": 소수점 둘째 자리까지 숫자 표시
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")       # fmt : format

plt.title("[시본] 피처 간 상관관계 히트맵")
plt.show()


# 실행 결과 설명
# 사용 데이터: 데이터프레임 내 모든 수치형 변수들

# 용도: 변수들 간 상관관계 탐색 및 변수 간 선형관계 강도 파악

# 해석:

# +1에 가까운 값은 강한 양의 상관관계, -1에 가까운 값은 강한 음의 상관관계를 의미

# 예를 들어, bmi와 target의 상관계수가 0.59 정도라면, BMI가 높을수록 당뇨 진행도(target)가 증가하는 경향이 있음을 시사

# 강한 상관관계 변수 쌍은 모델링 시 중요 변수 후보로 고려 가능

# 상관관계가 약하거나 0에 가까운 변수들은 독립적이라고 볼 수 있음
````

## 1.8 다중 플롯 (pairplot)

````python
# diag_kind='kde': 대각선에 각 변수의 커널 밀도 추정(KDE) 그래프를 표시 (히스토그램 대신 부드러운 분포 곡선)
sns.pairplot(df[['bmi', 's1', 'bp', 's5', 'target']], diag_kind='kde')

# pairplot은 자체적으로 Figure를 생성하므로 plt.suptitle()로 제목을 추가할 때 y 위치를 약간 위로 조정
plt.suptitle("[시본] 주요 변수 관계 시각화", y=1.02)

plt.show()


# 실행 결과 설명
# 사용 데이터: BMI, 혈압(bp), s5, 당뇨 진행도(target) 주요 수치형 변수

# 용도: 변수 간 쌍별 관계 탐색과 분포 형태 시각화

# 그래프 구성:

#   ①대각선: 각 변수별 KDE 분포곡선
#   ②비대각선: 변수 쌍별 산점도(Scatter plot)

# 해석:

#  ①산점도에서 점들이 일정한 선형 패턴을 보이면 양/음의 상관관계가 있음을 시사
#  ②BMI와 target 간에는 양의 상관관계가 보여 BMI가 클수록 당뇨 진행도가 높을 가능성 암시
#  ③대각선 KDE를 통해 각 변수의 분포 형태(정규성, 치우침 등)를 확인할 수 있음
#   ④변수 쌍 간 관계를 한눈에 파악할 수 있어 데이터 탐색(EDA)에 유용
````
