```python
# --------------------------------
# 1. 한글 CSV 안전하게 읽는 함수
# --------------------------------
def read_csv_korean(path):
    for enc in ["utf-8", "cp949", "euc-kr"]:
        try:
            return pd.read_csv(path, encoding=enc)
        except UnicodeDecodeError:
            continue
    raise UnicodeDecodeError(f"인코딩 실패: {path}")

# --------------------------------
# 2. 파일 경로
# --------------------------------
file_1 = "./source/시군구_성_월별_출생_202607.csv"

# --------------------------------
# 3. CSV 파일 읽기
# --------------------------------
df_kosis = read_csv_korean(file_1)

df_kosis.head()
```
